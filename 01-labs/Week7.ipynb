```python
sales_data = {
    "P001": {"name": "Widget A", "price": 9.99, "qty_sold": 120},
    "P002": {"name": "Widget B", "price": 14.50, "qty_sold": 85},
    "P003": {"name": "Widget C", "price": 7.25, "qty_sold": 200},
    "P004": {"name": "Widget D", "price": 12.75, "qty_sold": 75}
}

# Challenge 1: Total Revenue
total_revenue = sum(item["price"] * item["qty_sold"] for item in sales_data.values())
print(f"Total Revenue: ${total_revenue:.2f}")

# Challenge 2: Best-Selling Product
best_seller_id = max(sales_data, key=lambda p_id: sales_data[p_id]["qty_sold"])
best_product = sales_data[best_seller_id]
print(f"Best Seller: {best_product['name']} ({best_seller_id}) with {best_product['qty_sold']} units sold.")
```
